# Outils

**Exécutez cette cellule en premier.** `verifier` affiche ✅ ou ❌ ; `chrono_ms(f, *args)` donne le temps moyen en **millisecondes** sur 5 exécutions.

````python
def verifier(nom, obtenu, attendu):
    print(("✅" if obtenu == attendu else "❌"), nom, "| obtenu :", obtenu, "| attendu :", attendu)


import random
import sys
import time


def chrono_ms(f, *args, repetitions=5):
    # Temps moyen de f(*args) en millisecondes, sur 5 exécutions.
    total = 0.0
    for _ in range(repetitions):
        start = time.perf_counter()
        f(*args)
        total += time.perf_counter() - start
    return round(total / repetitions * 1000, 4)


print("Outils prêts ✅")
````

# Partie 3 — Équilibrage : arbre AVL

Si on insère des clés **dans l'ordre croissant**, l'ABR devient une liste : sa hauteur vaut $n - 1$ et toutes les opérations deviennent $O(n)$.

Un **arbre AVL** garde, pour chaque nœud, un **facteur d'équilibre** `hauteur(gauche) − hauteur(droite)` égal à −1, 0 ou 1. Après chaque insertion, s'il vaut ±2, on fait une **rotation** :

| Cas | Situation | Correction |
|---|---|---|
| LL | trop lourd à gauche, clé insérée à gauche du fils gauche | rotation droite |
| RR | trop lourd à droite, clé insérée à droite du fils droit | rotation gauche |
| LR | trop lourd à gauche, clé insérée à droite du fils gauche | rotation gauche sur le fils, puis droite |
| RL | trop lourd à droite, clé insérée à gauche du fils droit | rotation droite sur le fils, puis gauche |

> ⚠️ Dans le code AVL (comme dans la fiche), la hauteur se compte en **nœuds** : une feuille a une hauteur de **1**. Hauteur en arêtes = `node.height - 1`.

# Partie 4 — Mesures : ABR contre AVL

Pour $n = 100, 1\,000, 10\,000$ clés, on compare la **hauteur** (en arêtes) et le **temps total d'insertion** de trois arbres :
- un ABR rempli avec des clés **triées** (pire cas) ;
- un ABR rempli avec des clés **aléatoires** ;
- un AVL rempli avec des clés **triées**.

*(Exécutez seulement. L'ABR trié de 10 000 nœuds prend quelques secondes : c'est justement le pire cas.)*

````python
sys.setrecursionlimit(50_000)       # l'ABR trié est très profond
random.seed(4)

def avl_build(keys):
    r = None
    for k in keys:
        r = avl_insert(r, k)
    return r

print(f"{'n':>7} | {'ABR trié':>12} {'ABR aléat.':>12} {'AVL':>6} | {'temps ABR trié':>18} {'ABR aléat.':>12} {'AVL':>10}  (ms)")
for n in [100, 1_000, 10_000]:
    sorted_keys = list(range(n))
    random_keys = random.sample(range(10 * n), n)
    t0 = time.perf_counter(); worst = build(sorted_keys); t1 = time.perf_counter()
    rand = build(random_keys); t2 = time.perf_counter()
    avl = avl_build(sorted_keys); t3 = time.perf_counter()
    print(f"{n:>7,} | {str(worst.height(worst.root)):>12} {str(rand.height(rand.root)):>12} {h(avl) - 1:>6} | "
          f"{(t1 - t0) * 1000:>18.1f} {(t2 - t1) * 1000:>12.1f} {(t3 - t2) * 1000:>10.1f}")
````
