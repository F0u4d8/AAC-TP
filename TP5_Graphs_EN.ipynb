````python
import heapq
import time
from collections import deque


def check(name, got, expected):
    print(("✅" if got == expected else "❌"), name, "| got:", got, "| expected:", expected)


def time_ms(f, *args, repetitions=100):
    # Average time of f(*args) in milliseconds, over 100 runs.
    start = time.perf_counter()
    for _ in range(repetitions):
        f(*args)
    return (time.perf_counter() - start) / repetitions * 1000


print("Tools ready ✅")
````

# Part 3 — Graph traversals

**BFS (breadth-first)**: visit the source, then all its neighbours, then the neighbours' neighbours… with a **FIFO queue** (`deque`): `popleft()` removes from the front, `append()` adds at the back. A vertex is marked **visited when it is added** to the queue (it enters only once).

**DFS (depth-first)**: start from a vertex and go as far as possible before backtracking. **Recursive** version: mark `s` as visited, append it to the order, then call `dfs` on every neighbour not yet visited.

# Part 4 — Shortest paths: Dijkstra

**Principle:**
- initialise $d(A) = 0$ and $d(v) = \infty$ for $v \neq A$;
- at each iteration, pick the **unprocessed** vertex with the smallest distance (priority queue `heapq`);
- update the distances of its neighbours:

```
if d + w < dist[v]:
    dist[v] ← d + w ; pred[v] ← u ; push (dist[v], v) into the queue
```

`pred[v]` is the **predecessor** of `v` on the shortest path.

**Path reconstruction:** start from `destination`, walk back with `pred[current]` until `None`, then **reverse** the list. If the path does not start with `source` (unreachable vertex), return `None`.

## 🎯 Key takeaways

| Algorithm | Purpose | Key structure | Complexity |
|---|---|---|---|
| BFS | level-by-level traversal | FIFO queue (`deque`) | $O(V + E)$ |
| DFS | depth-first traversal | recursion (LIFO stack) | $O(V + E)$ |
| Dijkstra | shortest paths (weights ≥ 0) | priority queue (`heapq`) | $O((V + E) \log V)$ |

- Adjacency list of an undirected graph: each edge is stored **twice**.
- The visit order depends on the **order of neighbours** in the adjacency list.
- Dijkstra only works with **non-negative weights**; `pred` lets you **rebuild** the path.
