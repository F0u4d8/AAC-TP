**Run this cell first.** `check` prints ✅ or ❌; `timer_ms(f, *args)` gives the average time in **milliseconds** over 5 runs.

````python
def check(name, got, expected):
    print(("✅" if got == expected else "❌"), name, "| got :", got, "| expected :", expected)


import random
import sys
import time


def timer_ms(f, *args, repetitions=5):
    # Average time of f(*args) in milliseconds, over 5 runs.
    total = 0.0
    for _ in range(repetitions):
        start = time.perf_counter()
        f(*args)
        total += time.perf_counter() - start
    return round(total / repetitions * 1000, 4)


print("Tools ready ✅")
````

# Part 3 — Balancing: AVL tree

If keys are inserted **in increasing order**, the BST becomes a list: its height is $n - 1$ and every operation becomes $O(n)$.

An **AVL tree** keeps, for every node, a **balance factor** `height(left) − height(right)` equal to −1, 0 or 1. After each insertion, if it is ±2, we perform a **rotation**:

| Case | Situation | Fix |
|---|---|---|
| LL | left-heavy, key inserted left of the left child | right rotation |
| RR | right-heavy, key inserted right of the right child | left rotation |
| LR | left-heavy, key inserted right of the left child | left rotation on the child, then right |
| RL | right-heavy, key inserted left of the right child | right rotation on the child, then left |

> ⚠️ In the AVL code (as on the sheet), height is counted in **nodes**: a leaf has height **1**. Height in edges = `node.height - 1`.

# Part 4 — Measurements: BST vs AVL

For $n = 100, 1{,}000, 10{,}000$ keys, we compare the **height** (in edges) and the **total insertion time** of three trees:
- a BST filled with **sorted** keys (worst case);
- a BST filled with **random** keys;
- an AVL filled with **sorted** keys.

*(Just run it. The sorted BST with 10,000 nodes takes a few seconds: that is exactly the worst case.)*

````python
sys.setrecursionlimit(50_000)       # the sorted BST is very deep
random.seed(4)

def avl_build(keys):
    r = None
    for k in keys:
        r = avl_insert(r, k)
    return r

print(f"{'n':>7} | {'sorted BST':>12} {'random BST':>12} {'AVL':>6} | {'sorted BST time':>18} {'random BST':>12} {'AVL':>10}  (ms)")
for n in [100, 1_000, 10_000]:
    sorted_keys = list(range(n))
    random_keys = random.sample(range(10 * n), n)
    t0 = time.perf_counter(); worst = build(sorted_keys); t1 = time.perf_counter()
    rand = build(random_keys); t2 = time.perf_counter()
    avl = avl_build(sorted_keys); t3 = time.perf_counter()
    print(f"{n:>7,} | {str(worst.height(worst.root)):>12} {str(rand.height(rand.root)):>12} {h(avl) - 1:>6} | "
          f"{(t1 - t0) * 1000:>18.1f} {(t2 - t1) * 1000:>12.1f} {(t3 - t2) * 1000:>10.1f}")
````

## 🎯 Key takeaways

| Operation | BST | AVL |
|---|---|---|
| Search, insertion, deletion | $O(h)$: $O(\log n)$ if balanced, $O(n)$ worst case | $O(\log n)$ always |
| Traversal, height, node count | $O(n)$ | $O(n)$ |

- The **in-order** traversal of a BST gives the keys **sorted**.
- Deletion has **3 cases**: leaf, one child, two children (successor).
- AVL rotations (LL, RR, LR, RL) keep the height in $O(\log n)$.
